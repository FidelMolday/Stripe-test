```python
# ── Mount Google Drive ──────────────────────────────────────────────
from google.colab import drive
drive.mount('/content/drive')

# ── Install dependencies ────────────────────────────────────────────
!pip install imbalanced-learn xgboost -q

# ── Enable GPU: Runtime → Change runtime type → T4 GPU ─────────────
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Device: {device}")

import pandas as pd
import numpy as np
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.feature_selection import RFE
from sklearn.metrics import (accuracy_score, precision_score,
                             recall_score, f1_score, roc_auc_score,
                             classification_report)
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE, ADASYN
from imblearn.combine import SMOTEENN
from xgboost import XGBClassifier
import warnings
warnings.filterwarnings('ignore')

# ══════════════════════════════════════════════════════════════════
# 1. LOAD DATA
# ══════════════════════════════════════════════════════════════════
import glob

def load_cicids(path="/content/drive/MyDrive/IDS_Research/MachineLearningCVE/"):
    files = glob.glob(path + "*.csv")
    print(f"Found {len(files)} CICIDS files")
    df = pd.concat([pd.read_csv(f, encoding='utf-8',
                   low_memory=False) for f in files],
                   ignore_index=True)
    df.columns = df.columns.str.strip()
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(inplace=True)
    df.drop_duplicates(inplace=True)
    print(f"CICIDS loaded: {df.shape} | Classes: {df[' Label'].nunique()}")
    return df

def load_unsw(path="/content/drive/MyDrive/IDS_Research/UNSW/"):
    files = glob.glob(path + "*.csv")
    print(f"Found {len(files)} UNSW files")
    df = pd.concat([pd.read_csv(f, low_memory=False)
                   for f in files], ignore_index=True)
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(inplace=True)
    # Keep only feature columns + label
    if 'attack_cat' in df.columns:
        label_col = 'attack_cat'
    elif 'label' in df.columns:
        label_col = 'label'
    else:
        label_col = df.columns[-1]
    print(f"UNSW loaded: {df.shape} | Label col: {label_col}")
    return df, label_col

# ══════════════════════════════════════════════════════════════════
# 2. PREPROCESS
# ══════════════════════════════════════════════════════════════════
def preprocess(df, label_col, n_features=35):
    le = LabelEncoder()
    y  = le.fit_transform(df[label_col].astype(str).str.strip())
    X  = df.drop(columns=[label_col]).select_dtypes(include=[np.number])

    # Remove zero-variance columns
    X = X.loc[:, X.std() > 0]

    # RFE feature selection
    print("Running feature selection...")
    rf_sel = RandomForestClassifier(n_estimators=30,
                                    n_jobs=-1, random_state=42)
    n = min(n_features, X.shape[1])
    rfe = RFE(rf_sel, n_features_to_select=n, step=10)
    rfe.fit(X, y)
    X = X.loc[:, rfe.support_]

    # Normalise
    scaler = MinMaxScaler()
    X = scaler.fit_transform(X)
    print(f"Final shape: {X.shape} | Classes: {len(np.unique(y))}")
    return X, y, le.classes_

# ══════════════════════════════════════════════════════════════════
# 3. SPLIT 70 / 15 / 15
# ══════════════════════════════════════════════════════════════════
def split_data(X, y):
    X_tr, X_tmp, y_tr, y_tmp = train_test_split(
        X, y, test_size=0.30, stratify=y, random_state=42)
    X_val, X_te, y_val, y_te = train_test_split(
        X_tmp, y_tmp, test_size=0.50,
        stratify=y_tmp, random_state=42)
    print(f"Train: {X_tr.shape} | Val: {X_val.shape} | Test: {X_te.shape}")
    return X_tr, X_val, X_te, y_tr, y_val, y_te

# ══════════════════════════════════════════════════════════════════
# 4. IMBALANCE STRATEGIES
# ══════════════════════════════════════════════════════════════════
def apply_strategy(X_tr, y_tr, strategy):
    print(f"  Applying: {strategy}...")
    if strategy == "none":
        return X_tr, y_tr
    elif strategy == "smote":
        return SMOTE(random_state=42,
                     k_neighbors=3).fit_resample(X_tr, y_tr)
    elif strategy == "smoteenn":
        return SMOTEENN(random_state=42).fit_resample(X_tr, y_tr)
    elif strategy == "adasyn":
        return ADASYN(random_state=42,
                      n_neighbors=3).fit_resample(X_tr, y_tr)
    return X_tr, y_tr

# ══════════════════════════════════════════════════════════════════
# 5. TRANSFORMER-BiGRU
# ══════════════════════════════════════════════════════════════════
class TransformerBiGRU(nn.Module):
    def __init__(self, input_dim, d_model=128, nhead=4,
                 num_layers=2, d_ff=256, gru_hidden=64,
                 num_classes=2, dropout=0.1):
        super().__init__()
        self.embedding    = nn.Linear(input_dim, d_model)
        self.pos_emb      = nn.Parameter(torch.randn(1, 1, d_model))
        enc_layer         = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead,
            dim_feedforward=d_ff,
            dropout=dropout, batch_first=True)
        self.transformer  = nn.TransformerEncoder(
            enc_layer, num_layers=num_layers)
        self.bigru        = nn.GRU(
            d_model, gru_hidden,
            batch_first=True, bidirectional=True)
        self.classifier   = nn.Sequential(
            nn.Linear(gru_hidden * 2, 64),
            nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(64, num_classes))

    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.embedding(x) + self.pos_emb
        x = self.transformer(x)
        _, h = self.bigru(x)
        h = torch.cat([h[0], h[1]], dim=-1)
        return self.classifier(h)

# ══════════════════════════════════════════════════════════════════
# 6. CNN-IDS
# ══════════════════════════════════════════════════════════════════
class CNNIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Unflatten(1, (1, input_dim)),
            nn.Conv1d(1, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1), nn.Flatten(),
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x): return self.net(x)

# ══════════════════════════════════════════════════════════════════
# 7. LSTM-IDS
# ══════════════════════════════════════════════════════════════════
class LSTMIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 128, num_layers=2,
                            batch_first=True, dropout=0.2)
        self.head = nn.Sequential(
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        x = x.unsqueeze(1)
        _, (h, _) = self.lstm(x)
        return self.head(h[-1])

# ══════════════════════════════════════════════════════════════════
# 8. TRAIN & PREDICT HELPERS
# ══════════════════════════════════════════════════════════════════
def train_dl(model, X_tr, y_tr, X_val, y_val,
             epochs=30, batch_size=512,
             lr=1e-3, class_weights=None):
    model = model.to(device)
    Xtt   = torch.FloatTensor(X_tr).to(device)
    ytt   = torch.LongTensor(y_tr).to(device)
    Xvt   = torch.FloatTensor(X_val).to(device)
    yvt   = torch.LongTensor(y_val).to(device)

    loader = DataLoader(TensorDataset(Xtt, ytt),
                        batch_size=batch_size, shuffle=True)
    cw   = (torch.FloatTensor(class_weights).to(device)
            if class_weights is not None else None)
    loss_fn = nn.CrossEntropyLoss(weight=cw)
    opt     = torch.optim.Adam(model.parameters(), lr=lr)
    sched   = torch.optim.lr_scheduler.CosineAnnealingLR(
                  opt, T_max=epochs)

    best, patience, wait = np.inf, 5, 0
    for ep in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()
        sched.step()
        model.eval()
        with torch.no_grad():
            vl = loss_fn(model(Xvt), yvt).item()
        if vl < best:
            best = vl; wait = 0
        else:
            wait += 1
            if wait >= patience:
                break
    return model

def predict_dl(model, X_te):
    model.eval()
    with torch.no_grad():
        logits = model(torch.FloatTensor(X_te).to(device))
        probs  = torch.softmax(logits, -1).cpu().numpy()
    return np.argmax(probs, 1), probs

# ══════════════════════════════════════════════════════════════════
# 9. METRICS
# ══════════════════════════════════════════════════════════════════
def evaluate(y_true, y_pred, y_prob, multiclass=True):
    avg = 'macro'
    res = {
        "Accuracy":   round(accuracy_score(y_true, y_pred), 4),
        "Precision":  round(precision_score(
            y_true, y_pred, average=avg, zero_division=0), 4),
        "Recall":     round(recall_score(
            y_true, y_pred, average=avg, zero_division=0), 4),
        "F1_macro":   round(f1_score(
            y_true, y_pred, average='macro', zero_division=0), 4),
    }
    try:
        res["AUC_ROC"] = round(roc_auc_score(
            y_true, y_prob,
            multi_class='ovr', average='macro'), 4)
    except Exception:
        res["AUC_ROC"] = "N/A"
    return res

# ══════════════════════════════════════════════════════════════════
# 10. FULL EXPERIMENT
# ══════════════════════════════════════════════════════════════════
def run_all(X, y, dataset_name):
    num_classes = len(np.unique(y))
    print(f"\n{'═'*55}")
    print(f"  Dataset: {dataset_name} | Classes: {num_classes}")
    print(f"{'═'*55}")

    X_tr, X_val, X_te, y_tr, y_val, y_te = split_data(X, y)
    strategies = ["none","smote","smoteenn","adasyn","class_weights"]
    all_results = {}

    for strat in strategies:
        print(f"\n── Strategy: {strat} ──")

        if strat == "class_weights":
            X_s, y_s = X_tr, y_tr
            counts = np.bincount(y_tr)
            cw = len(y_tr) / (num_classes * counts.astype(float))
        else:
            X_s, y_s = apply_strategy(X_tr, y_tr, strat)
            cw = None

        strat_res = {}

        # Transformer-BiGRU
        m = TransformerBiGRU(X.shape[1],
                             num_classes=num_classes)
        m = train_dl(m, X_s, y_s, X_val, y_val,
                     class_weights=cw)
        p, pb = predict_dl(m, X_te)
        strat_res["Transformer-BiGRU"] = evaluate(y_te, p, pb)
        print(f"  Transformer-BiGRU: {strat_res['Transformer-BiGRU']}")

        # CNN-IDS
        m = CNNIDS(X.shape[1], num_classes)
        m = train_dl(m, X_s, y_s, X_val, y_val,
                     class_weights=cw)
        p, pb = predict_dl(m, X_te)
        strat_res["CNN-IDS"] = evaluate(y_te, p, pb)
        print(f"  CNN-IDS: {strat_res['CNN-IDS']}")

        # LSTM-IDS
        m = LSTMIDS(X.shape[1], num_classes)
        m = train_dl(m, X_s, y_s, X_val, y_val,
                     class_weights=cw)
        p, pb = predict_dl(m, X_te)
        strat_res["LSTM-IDS"] = evaluate(y_te, p, pb)
        print(f"  LSTM-IDS: {strat_res['LSTM-IDS']}")

        # Random Forest
        rf = RandomForestClassifier(
            n_estimators=200, class_weight='balanced',
            n_jobs=-1, random_state=42)
        rf.fit(X_s, y_s)
        p  = rf.predict(X_te)
        pb = rf.predict_proba(X_te)
        strat_res["Random Forest"] = evaluate(y_te, p, pb)
        print(f"  Random Forest: {strat_res['Random Forest']}")

        # XGBoost
        xgb = XGBClassifier(
            n_estimators=200, max_depth=6,
            learning_rate=0.1, subsample=0.8,
            use_label_encoder=False,
            eval_metric='mlogloss',
            n_jobs=-1, random_state=42)
        xgb.fit(X_s, y_s)
        p  = xgb.predict(X_te)
        pb = xgb.predict_proba(X_te)
        strat_res["XGBoost"] = evaluate(y_te, p, pb)
        print(f"  XGBoost: {strat_res['XGBoost']}")

        all_results[strat] = strat_res

    # ── Print final summary table ──────────────────────────────
    print(f"\n{'═'*55}")
    print(f"  FINAL RESULTS — {dataset_name} (SMOTE strategy)")
    print(f"{'═'*55}")
    print(f"{'Model':<22} {'Acc':>6} {'Prec':>6} "
          f"{'Rec':>6} {'F1':>6} {'AUC':>6}")
    print("-"*55)
    for model, metrics in all_results["smote"].items():
        print(f"  {model:<20} "
              f"{metrics['Accuracy']:>6} "
              f"{metrics['Precision']:>6} "
              f"{metrics['Recall']:>6} "
              f"{metrics['F1_macro']:>6} "
              f"{str(metrics['AUC_ROC']):>6}")

    return all_results

# ══════════════════════════════════════════════════════════════════
# 11. RUN EVERYTHING
# ══════════════════════════════════════════════════════════════════

# ── CICIDS-2017 ────────────────────────────────────────────────────
df_cic            = load_cicids()
X_cic, y_cic, _  = preprocess(df_cic, label_col=" Label")
results_cic       = run_all(X_cic, y_cic, "CICIDS-2017")

# ── UNSW-NB15 ──────────────────────────────────────────────────────
df_unsw, lbl      = load_unsw()
X_unsw, y_unsw, _ = preprocess(df_unsw, label_col=lbl)
results_unsw      = run_all(X_unsw, y_unsw, "UNSW-NB15")

print("\n✅ All experiments complete. Copy results into Section 4.")
```

```python
def load_cicids(path="/content/drive/MyDrive/IDS_Research/MachineLearningCSV/MachineLearningCVE/"):
    files = glob.glob(path + "*.csv")
    print(f"Found {len(files)} CICIDS files")

    dfs = []
    for f in files:
        df_temp = pd.read_csv(f, encoding='utf-8', low_memory=False)
        df_temp.columns = df_temp.columns.str.strip()
        df_temp.replace([np.inf, -np.inf], np.nan, inplace=True)
        df_temp.dropna(inplace=True)
        df_temp.drop_duplicates(inplace=True)

        # ── Stratified sample per file to preserve class balance ──
        # Keep max 50k rows per file, stratified by Label
        if len(df_temp) > 50000:
            # Sample proportionally per class
            df_temp = df_temp.groupby('Label', group_keys=False).apply(
                lambda x: x.sample(
                    min(len(x), max(10, int(50000 * len(x) / len(df_temp)))),
                    random_state=42))
        dfs.append(df_temp)

    df = pd.concat(dfs, ignore_index=True)
    print(f"CICIDS loaded: {df.shape} | Classes: {df['Label'].nunique()}")
    return df
```

```python
# ══════════════════════════════════════════════════════════════════
# COMPLETE IDS EXPERIMENT — TRANSFORMER-BiGRU vs BASELINES
# ══════════════════════════════════════════════════════════════════

# ── 1. LOAD ────────────────────────────────────────────────────────
def load_cicids():
    path = "/content/drive/MyDrive/IDS_Research/MachineLearningCSV/MachineLearningCVE/"
    files = glob.glob(path + "*.csv")
    print(f"Found {len(files)} CICIDS files")
    dfs = []
    for f in files:
        tmp = pd.read_csv(f, encoding='utf-8', low_memory=False)
        tmp.columns = tmp.columns.str.strip()
        tmp.replace([np.inf, -np.inf], np.nan, inplace=True)
        tmp.dropna(inplace=True)
        tmp.drop_duplicates(inplace=True)
        # Stratified sample — max 40k per file to save RAM
        if len(tmp) > 40000:
            tmp = tmp.groupby('Label', group_keys=False).apply(
                lambda x: x.sample(
                    min(len(x), max(10,
                    int(40000 * len(x) / len(tmp)))),
                    random_state=42))
        dfs.append(tmp)
        print(f"  Loaded: {f.split('/')[-1]} → {len(tmp)} rows")
    df = pd.concat(dfs, ignore_index=True)
    print(f"CICIDS total: {df.shape} | Classes: {df['Label'].nunique()}")
    return df

def load_unsw():
    train = "/content/drive/MyDrive/IDS_Research/UNSW/CSV Files/Training and Testing Sets/UNSW_NB15_training-set.csv"
    test  = "/content/drive/MyDrive/IDS_Research/UNSW/CSV Files/Training and Testing Sets/UNSW_NB15_testing-set.csv"
    df = pd.concat([pd.read_csv(train, low_memory=False),
                    pd.read_csv(test,  low_memory=False)],
                   ignore_index=True)
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(inplace=True)
    label_col = 'attack_cat' if 'attack_cat' in df.columns else 'label'
    # Clean label — strip whitespace, fill empty with 'Normal'
    df[label_col] = df[label_col].astype(str).str.strip()
    df[label_col] = df[label_col].replace('', 'Normal')
    print(f"UNSW total: {df.shape} | Classes: {df[label_col].nunique()}")
    return df, label_col

# ── 2. PREPROCESS ──────────────────────────────────────────────────
def preprocess(df, label_col, n_features=35):
    le = LabelEncoder()
    y  = le.fit_transform(df[label_col].astype(str).str.strip())
    X  = df.drop(columns=[label_col]).select_dtypes(include=[np.number])
    X  = X.loc[:, X.std() > 0]
    print(f"Running RFE on {X.shape[1]} features...")
    rf_sel = RandomForestClassifier(n_estimators=30,
                                    n_jobs=-1, random_state=42)
    n   = min(n_features, X.shape[1])
    rfe = RFE(rf_sel, n_features_to_select=n, step=10)
    rfe.fit(X, y)
    X = X.loc[:, rfe.support_]
    scaler = MinMaxScaler()
    X = scaler.fit_transform(X)
    print(f"Final: {X.shape} | Classes: {len(np.unique(y))}")
    return X, y, le.classes_

# ── 3. SPLIT ───────────────────────────────────────────────────────
def split_data(X, y):
    X_tr, X_tmp, y_tr, y_tmp = train_test_split(
        X, y, test_size=0.30, stratify=y, random_state=42)
    X_val, X_te, y_val, y_te = train_test_split(
        X_tmp, y_tmp, test_size=0.50,
        stratify=y_tmp, random_state=42)
    print(f"Train:{X_tr.shape} Val:{X_val.shape} Test:{X_te.shape}")
    return X_tr, X_val, X_te, y_tr, y_val, y_te

# ── 4. IMBALANCE ───────────────────────────────────────────────────
def apply_strategy(X_tr, y_tr, strategy):
    print(f"  Applying {strategy}...")
    if strategy == "none":
        return X_tr, y_tr
    elif strategy == "smote":
        return SMOTE(random_state=42,
                     k_neighbors=3).fit_resample(X_tr, y_tr)
    elif strategy == "smoteenn":
        return SMOTEENN(random_state=42).fit_resample(X_tr, y_tr)
    elif strategy == "adasyn":
        return ADASYN(random_state=42,
                      n_neighbors=3).fit_resample(X_tr, y_tr)
    return X_tr, y_tr

# ── 5. MODELS ──────────────────────────────────────────────────────
class TransformerBiGRU(nn.Module):
    def __init__(self, input_dim, num_classes,
                 d_model=128, nhead=4,
                 num_layers=2, d_ff=256,
                 gru_hidden=64, dropout=0.1):
        super().__init__()
        self.embedding   = nn.Linear(input_dim, d_model)
        self.pos_emb     = nn.Parameter(torch.randn(1, 1, d_model))
        enc              = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead,
            dim_feedforward=d_ff,
            dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(enc, num_layers)
        self.bigru       = nn.GRU(d_model, gru_hidden,
                                  batch_first=True,
                                  bidirectional=True)
        self.classifier  = nn.Sequential(
            nn.Linear(gru_hidden * 2, 64),
            nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(64, num_classes))
    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.embedding(x) + self.pos_emb
        x = self.transformer(x)
        _, h = self.bigru(x)
        return self.classifier(torch.cat([h[0], h[1]], dim=-1))

class CNNIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Unflatten(1, (1, input_dim)),
            nn.Conv1d(1, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1), nn.Flatten(),
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x): return self.net(x)

class LSTMIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 128, num_layers=2,
                            batch_first=True, dropout=0.2)
        self.head = nn.Sequential(
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        _, (h, _) = self.lstm(x.unsqueeze(1))
        return self.head(h[-1])

# ── 6. TRAIN / PREDICT ─────────────────────────────────────────────
def train_dl(model, X_tr, y_tr, X_val, y_val,
             epochs=30, batch_size=512,
             lr=1e-3, class_weights=None):
    model  = model.to(device)
    Xtt    = torch.FloatTensor(X_tr).to(device)
    ytt    = torch.LongTensor(y_tr).to(device)
    Xvt    = torch.FloatTensor(X_val).to(device)
    yvt    = torch.LongTensor(y_val).to(device)
    loader = DataLoader(TensorDataset(Xtt, ytt),
                        batch_size=batch_size, shuffle=True)
    cw      = (torch.FloatTensor(class_weights).to(device)
               if class_weights is not None else None)
    loss_fn = nn.CrossEntropyLoss(weight=cw)
    opt     = torch.optim.Adam(model.parameters(), lr=lr)
    sched   = torch.optim.lr_scheduler.CosineAnnealingLR(
                  opt, T_max=epochs)
    best, patience, wait = np.inf, 5, 0
    for ep in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()
        sched.step()
        model.eval()
        with torch.no_grad():
            vl = loss_fn(model(Xvt), yvt).item()
        if vl < best:
            best = vl; wait = 0
        else:
            wait += 1
            if wait >= patience: break
    return model

def predict_dl(model, X_te):
    model.eval()
    with torch.no_grad():
        logits = model(torch.FloatTensor(X_te).to(device))
        probs  = torch.softmax(logits, -1).cpu().numpy()
    return np.argmax(probs, 1), probs

# ── 7. METRICS ─────────────────────────────────────────────────────
def evaluate(y_true, y_pred, y_prob):
    res = {
        "Accuracy":  round(accuracy_score(y_true, y_pred), 4),
        "Precision": round(precision_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
        "Recall":    round(recall_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
        "F1_macro":  round(f1_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
    }
    try:
        res["AUC_ROC"] = round(roc_auc_score(
            y_true, y_prob,
            multi_class='ovr', average='macro'), 4)
    except Exception:
        res["AUC_ROC"] = "N/A"
    return res

# ── 8. RUN ALL ─────────────────────────────────────────────────────
def run_all(X, y, dataset_name):
    nc = len(np.unique(y))
    print(f"\n{'═'*55}\n  {dataset_name} | Classes: {nc}\n{'═'*55}")
    X_tr, X_val, X_te, y_tr, y_val, y_te = split_data(X, y)
    strategies = ["none","smote","smoteenn",
                  "adasyn","class_weights"]
    all_results = {}

    for strat in strategies:
        print(f"\n── Strategy: {strat} ──")
        if strat == "class_weights":
            X_s, y_s = X_tr, y_tr
            counts   = np.bincount(y_tr)
            cw       = len(y_tr) / (nc * counts.astype(float))
        else:
            X_s, y_s = apply_strategy(X_tr, y_tr, strat)
            cw       = None

        sr = {}

        # Transformer-BiGRU
        m = TransformerBiGRU(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["Transformer-BiGRU"] = evaluate(y_te, p, pb)
        print(f"  Transformer-BiGRU : {sr['Transformer-BiGRU']}")
        del m; torch.cuda.empty_cache()

        # CNN-IDS
        m = CNNIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["CNN-IDS"] = evaluate(y_te, p, pb)
        print(f"  CNN-IDS           : {sr['CNN-IDS']}")
        del m; torch.cuda.empty_cache()

        # LSTM-IDS
        m = LSTMIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["LSTM-IDS"] = evaluate(y_te, p, pb)
        print(f"  LSTM-IDS          : {sr['LSTM-IDS']}")
        del m; torch.cuda.empty_cache()

        # Random Forest
        rf = RandomForestClassifier(
            n_estimators=200, class_weight='balanced',
            n_jobs=-1, random_state=42)
        rf.fit(X_s, y_s)
        p  = rf.predict(X_te)
        pb = rf.predict_proba(X_te)
        sr["Random Forest"] = evaluate(y_te, p, pb)
        print(f"  Random Forest     : {sr['Random Forest']}")
        del rf

        # XGBoost
        xgb = XGBClassifier(
            n_estimators=200, max_depth=6,
            learning_rate=0.1, subsample=0.8,
            use_label_encoder=False,
            eval_metric='mlogloss',
            n_jobs=-1, random_state=42)
        xgb.fit(X_s, y_s)
        p  = xgb.predict(X_te)
        pb = xgb.predict_proba(X_te)
        sr["XGBoost"] = evaluate(y_te, p, pb)
        print(f"  XGBoost           : {sr['XGBoost']}")
        del xgb

        all_results[strat] = sr

    # ── Summary table ──────────────────────────────────────────
    print(f"\n{'═'*60}")
    print(f"  SUMMARY — {dataset_name}")
    print(f"{'═'*60}")
    for strat, models in all_results.items():
        print(f"\n  [{strat.upper()}]")
        print(f"  {'Model':<22} {'Acc':>6} {'Prec':>6} "
              f"{'Rec':>6} {'F1':>6} {'AUC':>7}")
        print(f"  {'-'*55}")
        for model, m in models.items():
            print(f"  {model:<22} "
                  f"{m['Accuracy']:>6} "
                  f"{m['Precision']:>6} "
                  f"{m['Recall']:>6} "
                  f"{m['F1_macro']:>6} "
                  f"{str(m['AUC_ROC']):>7}")
    return all_results

# ══════════════════════════════════════════════════════════════════
# EXECUTE
# ══════════════════════════════════════════════════════════════════
df_cic             = load_cicids()
X_cic, y_cic, _   = preprocess(df_cic, label_col="Label")
del df_cic
results_cic        = run_all(X_cic, y_cic, "CICIDS-2017")
del X_cic, y_cic

df_unsw, lbl       = load_unsw()
X_unsw, y_unsw, _  = preprocess(df_unsw, label_col=lbl)
del df_unsw
results_unsw       = run_all(X_unsw, y_unsw, "UNSW-NB15")
del X_unsw, y_unsw

print("\n✅ All experiments complete — paste results into Section 4.")
```

```python
# ══════════════════════════════════════════════════════════════════
# COMPLETE IDS EXPERIMENT — ONE CELL
# ══════════════════════════════════════════════════════════════════

# ── LOAD ───────────────────────────────────────────────────────────
def load_cicids():
    path  = "/content/drive/MyDrive/IDS_Research/MachineLearningCSV/MachineLearningCVE/"
    files = glob.glob(path + "*.csv")
    print(f"Found {len(files)} CICIDS files")
    dfs = []
    for f in files:
        tmp = pd.read_csv(f, encoding='utf-8', low_memory=False)
        tmp.columns = tmp.columns.str.strip()
        tmp.replace([np.inf, -np.inf], np.nan, inplace=True)
        tmp.dropna(inplace=True)
        tmp.drop_duplicates(inplace=True)
        if len(tmp) > 15000:
            tmp = tmp.groupby('Label', group_keys=False).apply(
                lambda x: x.sample(
                    min(len(x), max(10,
                    int(15000 * len(x) / len(tmp)))),
                    random_state=42))
        dfs.append(tmp)
        print(f"  {f.split('/')[-1]} → {len(tmp)} rows")
    df = pd.concat(dfs, ignore_index=True)
    print(f"CICIDS total: {df.shape} | Classes: {df['Label'].nunique()}")
    return df

def load_unsw():
    train = "/content/drive/MyDrive/IDS_Research/UNSW/CSV Files/Training and Testing Sets/UNSW_NB15_training-set.csv"
    test  = "/content/drive/MyDrive/IDS_Research/UNSW/CSV Files/Training and Testing Sets/UNSW_NB15_testing-set.csv"
    df    = pd.concat([pd.read_csv(train, low_memory=False),
                       pd.read_csv(test,  low_memory=False)],
                      ignore_index=True)
    df.replace([np.inf, -np.inf], np.nan, inplace=True)
    df.dropna(inplace=True)
    label_col = 'attack_cat' if 'attack_cat' in df.columns else 'label'
    df[label_col] = df[label_col].astype(str).str.strip()
    df[label_col] = df[label_col].replace('', 'Normal')
    print(f"UNSW total: {df.shape} | Classes: {df[label_col].nunique()}")
    return df, label_col

# ── PREPROCESS ─────────────────────────────────────────────────────
def preprocess(df, label_col, n_features=35):
    le = LabelEncoder()
    y  = le.fit_transform(df[label_col].astype(str).str.strip())
    X  = df.drop(columns=[label_col]).select_dtypes(include=[np.number])
    X  = X.loc[:, X.std() > 0]
    print(f"Running RFE on {X.shape[1]} features...")
    idx    = np.random.choice(len(X), min(20000, len(X)), replace=False)
    rf_sel = RandomForestClassifier(n_estimators=20,
                                    n_jobs=-1, random_state=42)
    rfe    = RFE(rf_sel,
                 n_features_to_select=min(n_features, X.shape[1]),
                 step=10)
    rfe.fit(X.iloc[idx], y[idx])
    X      = X.loc[:, rfe.support_]
    scaler = MinMaxScaler()
    X      = scaler.fit_transform(X)
    print(f"Final: {X.shape} | Classes: {len(np.unique(y))}")
    return X, y, le.classes_

# ── SPLIT ──────────────────────────────────────────────────────────
def split_data(X, y):
    X_tr, X_tmp, y_tr, y_tmp = train_test_split(
        X, y, test_size=0.30, stratify=y, random_state=42)
    X_val, X_te, y_val, y_te = train_test_split(
        X_tmp, y_tmp, test_size=0.50,
        stratify=y_tmp, random_state=42)
    print(f"Train:{X_tr.shape} Val:{X_val.shape} Test:{X_te.shape}")
    return X_tr, X_val, X_te, y_tr, y_val, y_te

# ── IMBALANCE ──────────────────────────────────────────────────────
def apply_strategy(X_tr, y_tr, strategy):
    print(f"  Applying {strategy}...")
    if strategy == "none":
        return X_tr, y_tr
    elif strategy == "smote":
        return SMOTE(random_state=42,
                     k_neighbors=3).fit_resample(X_tr, y_tr)
    elif strategy == "smoteenn":
        return SMOTEENN(random_state=42).fit_resample(X_tr, y_tr)
    elif strategy == "adasyn":
        return ADASYN(random_state=42,
                      n_neighbors=3).fit_resample(X_tr, y_tr)
    return X_tr, y_tr

# ── MODELS ─────────────────────────────────────────────────────────
class TransformerBiGRU(nn.Module):
    def __init__(self, input_dim, num_classes,
                 d_model=128, nhead=4, num_layers=2,
                 d_ff=256, gru_hidden=64, dropout=0.1):
        super().__init__()
        self.embedding   = nn.Linear(input_dim, d_model)
        self.pos_emb     = nn.Parameter(torch.randn(1, 1, d_model))
        enc              = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead,
            dim_feedforward=d_ff,
            dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(enc, num_layers)
        self.bigru       = nn.GRU(d_model, gru_hidden,
                                  batch_first=True, bidirectional=True)
        self.classifier  = nn.Sequential(
            nn.Linear(gru_hidden * 2, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.embedding(x) + self.pos_emb
        x = self.transformer(x)
        _, h = self.bigru(x)
        return self.classifier(torch.cat([h[0], h[1]], dim=-1))

class CNNIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Unflatten(1, (1, input_dim)),
            nn.Conv1d(1, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1), nn.Flatten(),
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x): return self.net(x)

class LSTMIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 128, num_layers=2,
                            batch_first=True, dropout=0.2)
        self.head = nn.Sequential(
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        _, (h, _) = self.lstm(x.unsqueeze(1))
        return self.head(h[-1])

# ── TRAIN / PREDICT ────────────────────────────────────────────────
def train_dl(model, X_tr, y_tr, X_val, y_val,
             epochs=30, batch_size=512,
             lr=1e-3, class_weights=None):
    model   = model.to(device)
    Xtt     = torch.FloatTensor(X_tr).to(device)
    ytt     = torch.LongTensor(y_tr).to(device)
    Xvt     = torch.FloatTensor(X_val).to(device)
    yvt     = torch.LongTensor(y_val).to(device)
    loader  = DataLoader(TensorDataset(Xtt, ytt),
                         batch_size=batch_size, shuffle=True)
    cw      = (torch.FloatTensor(class_weights).to(device)
               if class_weights is not None else None)
    loss_fn = nn.CrossEntropyLoss(weight=cw)
    opt     = torch.optim.Adam(model.parameters(), lr=lr)
    sched   = torch.optim.lr_scheduler.CosineAnnealingLR(
                  opt, T_max=epochs)
    best, patience, wait = np.inf, 5, 0
    for ep in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()
        sched.step()
        model.eval()
        with torch.no_grad():
            vl = loss_fn(model(Xvt), yvt).item()
        if vl < best:
            best = vl; wait = 0
        else:
            wait += 1
            if wait >= patience: break
    return model

def predict_dl(model, X_te):
    model.eval()
    with torch.no_grad():
        logits = model(torch.FloatTensor(X_te).to(device))
        probs  = torch.softmax(logits, -1).cpu().numpy()
    return np.argmax(probs, 1), probs

# ── METRICS ────────────────────────────────────────────────────────
def evaluate(y_true, y_pred, y_prob):
    res = {
        "Accuracy":  round(accuracy_score(y_true, y_pred), 4),
        "Precision": round(precision_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
        "Recall":    round(recall_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
        "F1_macro":  round(f1_score(
            y_true, y_pred, average='macro',
            zero_division=0), 4),
    }
    try:
        res["AUC_ROC"] = round(roc_auc_score(
            y_true, y_prob,
            multi_class='ovr', average='macro'), 4)
    except Exception:
        res["AUC_ROC"] = "N/A"
    return res

# ── RUN ALL ────────────────────────────────────────────────────────
def run_all(X, y, dataset_name):
    nc = len(np.unique(y))
    print(f"\n{'═'*55}\n  {dataset_name} | Classes: {nc}\n{'═'*55}")
    X_tr, X_val, X_te, y_tr, y_val, y_te = split_data(X, y)
    strategies  = ["none","smote","smoteenn","adasyn","class_weights"]
    all_results = {}

    for strat in strategies:
        print(f"\n── Strategy: {strat} ──")
        if strat == "class_weights":
            X_s, y_s = X_tr, y_tr
            counts   = np.bincount(y_tr)
            cw       = len(y_tr) / (nc * counts.astype(float))
        else:
            X_s, y_s = apply_strategy(X_tr, y_tr, strat)
            cw       = None

        sr = {}

        m = TransformerBiGRU(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["Transformer-BiGRU"] = evaluate(y_te, p, pb)
        print(f"  Transformer-BiGRU : {sr['Transformer-BiGRU']}")
        del m; torch.cuda.empty_cache(); gc.collect()

        m = CNNIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["CNN-IDS"] = evaluate(y_te, p, pb)
        print(f"  CNN-IDS           : {sr['CNN-IDS']}")
        del m; torch.cuda.empty_cache(); gc.collect()

        m = LSTMIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["LSTM-IDS"] = evaluate(y_te, p, pb)
        print(f"  LSTM-IDS          : {sr['LSTM-IDS']}")
        del m; torch.cuda.empty_cache(); gc.collect()

        rf = RandomForestClassifier(
            n_estimators=200, class_weight='balanced',
            n_jobs=-1, random_state=42)
        rf.fit(X_s, y_s)
        p  = rf.predict(X_te)
        pb = rf.predict_proba(X_te)
        sr["Random Forest"] = evaluate(y_te, p, pb)
        print(f"  Random Forest     : {sr['Random Forest']}")
        del rf; gc.collect()

        xgb = XGBClassifier(
            n_estimators=200, max_depth=6,
            learning_rate=0.1, subsample=0.8,
            use_label_encoder=False,
            eval_metric='mlogloss',
            n_jobs=-1, random_state=42)
        xgb.fit(X_s, y_s)
        p  = xgb.predict(X_te)
        pb = xgb.predict_proba(X_te)
        sr["XGBoost"] = evaluate(y_te, p, pb)
        print(f"  XGBoost           : {sr['XGBoost']}")
        del xgb; gc.collect()

        all_results[strat] = sr

    # ── Summary ────────────────────────────────────────────────
    print(f"\n{'═'*60}")
    print(f"  FINAL SUMMARY — {dataset_name}")
    print(f"{'═'*60}")
    for strat, models in all_results.items():
        print(f"\n  [{strat.upper()}]")
        print(f"  {'Model':<22} {'Acc':>6} {'Prec':>6} "
              f"{'Rec':>6} {'F1':>6} {'AUC':>7}")
        print(f"  {'-'*55}")
        for model, m in models.items():
            print(f"  {model:<22} "
                  f"{m['Accuracy']:>6} "
                  f"{m['Precision']:>6} "
                  f"{m['Recall']:>6} "
                  f"{m['F1_macro']:>6} "
                  f"{str(m['AUC_ROC']):>7}")
    return all_results

# ══════════════════════════════════════════════════════════════════
# EXECUTE
# ══════════════════════════════════════════════════════════════════
df_cic            = load_cicids()
X_cic, y_cic, _  = preprocess(df_cic, label_col="Label")
del df_cic; gc.collect()
results_cic       = run_all(X_cic, y_cic, "CICIDS-2017")
del X_cic, y_cic; gc.collect()

df_unsw, lbl      = load_unsw()
X_unsw, y_unsw, _ = preprocess(df_unsw, label_col=lbl)
del df_unsw; gc.collect()
results_unsw      = run_all(X_unsw, y_unsw, "UNSW-NB15")
del X_unsw, y_unsw; gc.collect()

print("\n✅ All experiments complete — paste results into Section 4.")
```

```python
class TransformerBiGRU(nn.Module):
    def __init__(self, input_dim, num_classes,
                 d_model=128, nhead=4, num_layers=2,
                 d_ff=256, gru_hidden=64, dropout=0.1):
        super().__init__()
        self.embedding   = nn.Linear(input_dim, d_model)
        self.pos_emb     = nn.Parameter(torch.randn(1, 1, d_model))
        enc              = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=nhead,
            dim_feedforward=d_ff,
            dropout=dropout, batch_first=True)
        self.transformer = nn.TransformerEncoder(enc, num_layers)
        self.bigru       = nn.GRU(d_model, gru_hidden,
                                  batch_first=True, bidirectional=True)
        self.classifier  = nn.Sequential(
            nn.Linear(gru_hidden * 2, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.embedding(x) + self.pos_emb
        x = self.transformer(x)
        _, h = self.bigru(x)
        return self.classifier(torch.cat([h[0], h[1]], dim=-1))

class CNNIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.net = nn.Sequential(
            nn.Unflatten(1, (1, input_dim)),
            nn.Conv1d(1, 64, 3, padding=1),
            nn.BatchNorm1d(64), nn.ReLU(),
            nn.Conv1d(64, 128, 3, padding=1),
            nn.BatchNorm1d(128), nn.ReLU(),
            nn.AdaptiveAvgPool1d(1), nn.Flatten(),
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x): return self.net(x)

class LSTMIDS(nn.Module):
    def __init__(self, input_dim, num_classes):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, 128, num_layers=2,
                            batch_first=True, dropout=0.2)
        self.head = nn.Sequential(
            nn.Linear(128, 64), nn.ReLU(),
            nn.Dropout(0.3), nn.Linear(64, num_classes))
    def forward(self, x):
        _, (h, _) = self.lstm(x.unsqueeze(1))
        return self.head(h[-1])

def train_dl(model, X_tr, y_tr, X_val, y_val,
             epochs=20, batch_size=1024,
             lr=1e-3, class_weights=None):
    model   = model.to(device)
    Xtt     = torch.FloatTensor(X_tr).to(device)
    ytt     = torch.LongTensor(y_tr).to(device)
    Xvt     = torch.FloatTensor(X_val).to(device)
    yvt     = torch.LongTensor(y_val).to(device)
    loader  = DataLoader(TensorDataset(Xtt, ytt),
                         batch_size=batch_size, shuffle=True)
    cw      = (torch.FloatTensor(class_weights).to(device)
               if class_weights is not None else None)
    loss_fn = nn.CrossEntropyLoss(weight=cw)
    opt     = torch.optim.Adam(model.parameters(), lr=lr)
    sched   = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    best, patience, wait = np.inf, 3, 0
    for ep in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()
        sched.step()
        model.eval()
        with torch.no_grad():
            vl = loss_fn(model(Xvt), yvt).item()
        print(f"    ep{ep+1} val_loss={vl:.4f}")
        if vl < best:
            best = vl; wait = 0
        else:
            wait += 1
            if wait >= patience: break
    return model

def predict_dl(model, X_te):
    model.eval()
    with torch.no_grad():
        logits = model(torch.FloatTensor(X_te).to(device))
        probs  = torch.softmax(logits, -1).cpu().numpy()
    return np.argmax(probs, 1), probs

def evaluate(y_true, y_pred, y_prob):
    res = {
        "Accuracy":  round(accuracy_score(y_true, y_pred), 4),
        "Precision": round(precision_score(y_true, y_pred,
            average='macro', zero_division=0), 4),
        "Recall":    round(recall_score(y_true, y_pred,
            average='macro', zero_division=0), 4),
        "F1_macro":  round(f1_score(y_true, y_pred,
            average='macro', zero_division=0), 4),
    }
    try:
        res["AUC_ROC"] = round(roc_auc_score(
            y_true, y_prob, multi_class='ovr', average='macro'), 4)
    except Exception:
        res["AUC_ROC"] = "N/A"
    return res

def run_all_lite(X, y, dataset_name):
    nc = len(np.unique(y))
    print(f"\n{'═'*55}\n  {dataset_name} | Classes: {nc}\n{'═'*55}")
    X_tr, X_val, X_te, y_tr, y_val, y_te = split_data(X, y)
    strategies  = ["none", "smote", "adasyn", "class_weights"]
    all_results = {}
    for strat in strategies:
        print(f"\n── Strategy: {strat} ──")
        if strat == "class_weights":
            X_s, y_s = X_tr, y_tr
            counts   = np.bincount(y_tr)
            cw       = len(y_tr) / (nc * counts.astype(float))
        else:
            X_s, y_s = apply_strategy(X_tr, y_tr, strat)
            cw       = None
        sr = {}

        print("  Training Transformer-BiGRU...")
        m = TransformerBiGRU(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["Transformer-BiGRU"] = evaluate(y_te, p, pb)
        print(f"  ✅ Transformer-BiGRU : {sr['Transformer-BiGRU']}")
        del m; gc.collect()

        print("  Training CNN-IDS...")
        m = CNNIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["CNN-IDS"] = evaluate(y_te, p, pb)
        print(f"  ✅ CNN-IDS           : {sr['CNN-IDS']}")
        del m; gc.collect()

        print("  Training LSTM-IDS...")
        m = LSTMIDS(X.shape[1], nc)
        m = train_dl(m, X_s, y_s, X_val, y_val, class_weights=cw)
        p, pb = predict_dl(m, X_te)
        sr["LSTM-IDS"] = evaluate(y_te, p, pb)
        print(f"  ✅ LSTM-IDS          : {sr['LSTM-IDS']}")
        del m; gc.collect()

        print("  Training Random Forest...")
        rf = RandomForestClassifier(n_estimators=200,
            class_weight='balanced', n_jobs=-1, random_state=42)
        rf.fit(X_s, y_s)
        p  = rf.predict(X_te)
        pb = rf.predict_proba(X_te)
        sr["Random Forest"] = evaluate(y_te, p, pb)
        print(f"  ✅ Random Forest     : {sr['Random Forest']}")
        del rf; gc.collect()

        print("  Training XGBoost...")
        xgb = XGBClassifier(n_estimators=200, max_depth=6,
            learning_rate=0.1, subsample=0.8,
            use_label_encoder=False, eval_metric='mlogloss',
            n_jobs=-1, random_state=42)
        xgb.fit(X_s, y_s)
        p  = xgb.predict(X_te)
        pb = xgb.predict_proba(X_te)
        sr["XGBoost"] = evaluate(y_te, p, pb)
        print(f"  ✅ XGBoost           : {sr['XGBoost']}")
        del xgb; gc.collect()

        all_results[strat] = sr

    print(f"\n{'═'*60}\n  FINAL SUMMARY — {dataset_name}\n{'═'*60}")
    for strat, models in all_results.items():
        print(f"\n  [{strat.upper()}]")
        print(f"  {'Model':<22} {'Acc':>6} {'Prec':>6} "
              f"{'Rec':>6} {'F1':>6} {'AUC':>7}")
        print(f"  {'-'*55}")
        for model, m in models.items():
            print(f"  {model:<22} "
                  f"{m['Accuracy']:>6} "
                  f"{m['Precision']:>6} "
                  f"{m['Recall']:>6} "
                  f"{m['F1_macro']:>6} "
                  f"{str(m['AUC_ROC']):>7}")
    return all_results

# ── EXECUTE ────────────────────────────────────────────────────────
df_cic            = load_cicids()
X_cic, y_cic, _  = preprocess(df_cic, label_col="Label")
del df_cic; gc.collect()
results_cic       = run_all_lite(X_cic, y_cic, "CICIDS-2017")
del X_cic, y_cic; gc.collect()

df_unsw, lbl      = load_unsw()
X_unsw, y_unsw, _ = preprocess(df_unsw, label_col=lbl)
del df_unsw; gc.collect()
results_unsw      = run_all_lite(X_unsw, y_unsw, "UNSW-NB15")
del X_unsw, y_unsw; gc.collect()

print("\n✅ All experiments complete — paste results into Section 4.")
```
